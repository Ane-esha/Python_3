```python
import random
secret_number = random.randint(1, 10)

print("Guess a number between 1 and 10")

# Maximum number of attempts
attempts = 3

while attempts > 0:
    guess = int(input("Guess the number (1 to 10): "))

    # Check if guess is outside the valid range
    if guess < 1 or guess > 10:
        print("Please enter a number between 1 and 10.")
        continue

    # Check if guess is correct
    if guess == secret_number:
        print("Congratulations! You guessed the correct number.")
        break

    # Check if guess is greater
    elif guess > secret_number:
        print("Your guess is too high.")

    # Check if guess is lower
    else:
        print("Your guess is too low.")

    attempts = attempts - 1

else:
    print("Game over! You have used all your attempts.")
    print("The secret number was:", secret_number)
```

```python

num = int(input("Enter the number for which you want the multiplication table: "))

for i in range(1, 11):
    result = num * i
    print(num, "x", i, "=", result)
```

```python
def calculate_bmi(weight, height):
    bmi = weight / (height ** 2)
    return bmi

weight = float(input("Enter your weight in kg: "))
height = float(input("Enter your height in meters: "))

bmi = calculate_bmi(weight, height)

print("Your BMI is:", round(bmi,2))
```
